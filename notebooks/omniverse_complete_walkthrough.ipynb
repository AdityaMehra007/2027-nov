## 1. Quantitative Finance: Indicators, Vectorized Backtesting & Monte Carlo VaR
We evaluate technical momentum indicators (RSI, MACD, Bollinger Bands, ATR), run moving average crossover backtesting, and simulate 1,000 multi-path Monte Carlo trajectories to compute parametric Value-at-Risk ($VaR_{95}, VaR_{99}$) and Expected Shortfall ($CVaR$).

```python
from omniverse.modules.intelligence.quant_engine import Backtester
from omniverse.modules.intelligence.indicators import (
    calculate_rsi,
    calculate_macd,
    calculate_bollinger_bands,
    calculate_atr
)

prices = [23800.0, 23950.0, 23890.0, 24100.0, 24250.0, 24180.0, 24350.0, 
          24480.0, 24420.0, 24600.0, 24750.0, 24900.0, 24820.0, 25050.0, 25200.0]

rsi = calculate_rsi(prices)
macd = calculate_macd(prices)
bb = calculate_bollinger_bands(prices)
print(f"Latest Index Price: {prices[-1]:,.2f}")
print(f"RSI(14): {rsi:.2f}")
print(f"MACD Histogram: {macd['histogram']:.2f}")
print(f"Bollinger Bands: [{bb['lower_band']:,.2f}, {bb['upper_band']:,.2f}]")

backtester = Backtester()
var_res = backtester.run_monte_carlo_var(prices, num_simulations=1000, horizon_days=30)
print(f"Monte Carlo VaR (95%): {var_res['var_95_pct']}%")
print(f"Monte Carlo VaR (99%): {var_res['var_99_pct']}%")
print(f"Expected Shortfall (CVaR): {var_res['expected_shortfall_cvar_pct']}%")
```

## 2. Options Derivatives Greeks & Cross-Currency Triangular Arbitrage
Compute closed-form Black-Scholes Greeks ($\Delta, \Gamma, \Theta, \mathcal{V}, \rho$) and scan foreign exchange / cryptocurrency cycles for negative log-weight triangular arbitrage.

```python
from omniverse.modules.intelligence.derivatives import DerivativesEngine

engine = DerivativesEngine()
greeks = engine.calculate_black_scholes_greeks(
    spot=100.0,
    strike=105.0,
    time_to_maturity_years=0.5,
    risk_free_rate=0.05,
    volatility=0.22
)
print("--- Black-Scholes Option Greeks ---")
print(f"Call Price: ${greeks.call_price:.2f} | Put Price: ${greeks.put_price:.2f}")
print(f"Delta Call: {greeks.delta_call:.4f} | Delta Put: {greeks.delta_put:.4f} | Gamma: {greeks.gamma:.4f}")
print(f"Theta Call: {greeks.theta_call:.4f} | Vega: {greeks.vega:.4f} | Rho Call: {greeks.rho_call:.4f}")

rates = {"USD_INR": 83.50, "EUR_USD": 1.085, "EUR_INR": 91.20}
arb = engine.detect_triangular_arbitrage(rates=rates, initial_capital=100_000.0)
print("\n--- Triangular Arbitrage Detection ---")
print(f"Cycle Path: {' -> '.join(arb.pair_path)}")
print(f"Initial: {arb.start_amount:,.2f} -> Final: {arb.end_amount:,.2f} | Profit: {arb.profit_pct:.4f}%")
print(f"Arbitrage Detected (>15bps): {arb.arbitrage_detected}")
```

## 4. Sentinel-1 SAR Radar Processing & Hazard Inundation Mapping
Orbital radar microwave backscatter ($\sigma^0$) is processed using 2D speckle smoothing and Otsu bimodal valley thresholding to extract water body inundation polygons.

```python
from omniverse.modules.aether.sar_processor import SARProcessor

sar = SARProcessor()
# Synthetic swath: water pixels (-22 to -25 dB) and land pixels (-8 to -12 dB)
backscatter_grid = [-23.5, -22.0, -24.1, -21.8, -12.4, -9.1, -8.3, -11.0] * 8
mask = sar.process_sar_swath(area_name="Brahmaputra_Basin_Sector_A", backscatter_db_grid=backscatter_grid, pixel_resolution_m=10.0)
print(f"AOI: {mask.target_area}")
print(f"Total Sample Pixels: {mask.total_pixels}")
print(f"Inundated Water Pixels: {mask.water_pixels}")
print(f"Calculated Otsu Threshold: {mask.optimal_otsu_threshold_db:.2f} dB")
print(f"Flooded Area: {mask.inundation_area_km2:.4f} km^2")
print(f"Confidence Score: {mask.confidence_score:.2f}")
```
